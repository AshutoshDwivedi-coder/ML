## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

I will choose Clssification. I want to classify content into declining or not declining based on search-performance and query-related features. Classification fits because the outcome has two categories and the model can help prioritize pages that may need attention.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

Target: whether a content item experiences a significant decline in search impressions. The label is created from an observed performance rule: a page is marked as declining when its recent impressions are less than 80% of its previous-period impressions. This is a defined proxy based on measured search performance rather than a manually assigned label.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

Precision. Precision measures how often pages predicted as declining are actually declining. A higher precision means fewer healthy pages are incorrectly prioritized for action. I would consider a precision above 0.90 useful because it keeps false-positive recommendations relatively low.

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

One row represents one content item (page) for a client, with its search-performance and query-related features. The row contains information such as previous-period impressions, clicks, query signals, position volatility, and the target indicating whether the content declined.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

The pattern is too complex for a single if-statement because content performance depends on multiple signals at the same time. Impressions, query concentration, position volatility, and other features can interact in different ways, so ML can learn combinations of signals instead of relying on one fixed threshold.
